```python
def classify_control(description):
    text = description.lower()

    if any(word in text for word in ["policy", "management", "responsibility"]):
        return "Organizational Controls"
    elif any(word in text for word in ["employee", "training", "awareness"]):
        return "People Controls"
    elif any(word in text for word in ["door", "camera", "physical", "building"]):
        return "Physical Controls"
    elif any(word in text for word in ["firewall", "encryption", "network", "access"]):
        return "Technological Controls"
    else:
        return "Control Domain Not Identified"

description = input("Enter security control description: ")
print("Control Domain:", classify_control(description))
```
